## 1. Pemahaman Kasus & Tipe Data

Gambar tugas menunjukkan dataset operasional penjualan dengan **5 kolom** dan 2 baris contoh. Tiap kolom sudah dipetakan ke skala pengukuran dan jenis visualisasi yang cocok:

| Kolom | Contoh | Tipe / Skala | Alasan | Visualisasi yang sesuai |
|---|---|---|---|---|
| **Produk** | Laptop, Buku | Kategorik - Nominal | Label tanpa urutan | Bar chart / Pie chart |
| **Kategori** | Elektronik, Pendidikan | Kategorik - Nominal | Label tanpa urutan | Bar chart / Pie chart |

**Konvensi warna** di seluruh notebook mengikuti gambar tugas: **biru** = kategorik, **oranye** = numerik, **abu-abu** = waktu.

## 2. Persiapan Lingkungan

````python
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.ticker import FuncFormatter
import seaborn as sns
from scipy import stats
from IPython.display import display, Markdown

# ----- Identitas (wajib) -----
NAMA, NIM = "Ganesia Alkhan Fikriano", "241051019"

# ----- Warna sesuai gambar tugas -----
BIRU, ORANYE, ABU = "#2E9BD6", "#F5A03A", "#7A7F87"
PALET_KAT = sns.color_palette("Blues_r", 9)[:7]

sns.set_theme(style="whitegrid", context="notebook")
plt.rcParams.update({"figure.dpi": 110, "axes.titleweight": "bold", "axes.titlesize": 11.5,
                     "axes.labelsize": 10, "figure.facecolor": "#FBF8EF", "savefig.facecolor": "#FBF8EF"})
pd.set_option("display.max_columns", 30)
pd.set_option("display.float_format", lambda v: f"{v:,.2f}")

HARGA, QTY = "Harga (Rp)", "Jumlah Terjual"
NAMA_BULAN = ["Jan", "Feb", "Mar", "Apr", "Mei", "Jun", "Jul", "Agu", "Sep", "Okt", "Nov", "Des"]
NAMA_HARI = ["Senin", "Selasa", "Rabu", "Kamis", "Jumat", "Sabtu", "Minggu"]


def rp(x):
    """Format rupiah gaya Indonesia: Rp 8.500.000"""
    return "Rp " + f"{x:,.0f}".replace(",", ".")


def fmt_juta(v, _=None):
    return f"{v/1e6:g} jt"


def fmt_rupiah_singkat(v, _=None):
    return f"{v/1e6:g} jt" if v >= 1e6 else f"{v/1e3:g} rb"


def selesai(fig, judul=None):
    """Tampilkan figure dengan judul dan identitas mahasiswa di setiap grafik."""
    if judul:
        fig.suptitle(judul, fontsize=14, fontweight="bold")
    fig.text(0.995, 0.006, f"{NAMA} | NIM {NIM}", ha="right", va="bottom", fontsize=8, color="#666666")
    fig.tight_layout(rect=[0, 0.03, 1, 0.95 if judul else 1])
    plt.show()


display(Markdown(f"**Pemilik notebook:** {NAMA} &nbsp;|&nbsp; **NIM:** {NIM}"))
````

## 3. Pembuatan Data Sintetis (1.000 baris)

Agar data **realistis**, generator dirancang dengan aturan berikut:

- **Katalog 36 produk dalam 7 kategori** dengan harga dasar yang masuk akal (Laptop ~Rp 8,5 juta, Buku Pelajaran ~Rp 75 ribu, Keripik ~Rp 20 ribu).
- **Harga** berfluktuasi di sekitar harga dasar (+/- sekitar 6%), dibulatkan sesuai kebiasaan harga di Indonesia, dan sesekali terkena **promo** 8-20% (lebih sering pada tanggal kembar seperti 11.11 dan 12.12).
- **Jumlah terjual** (bilangan bulat) mengikuti distribusi Poisson: barang murah laku lebih banyak, barang mahal lebih sedikit, dan harga promo menaikkan penjualan (elastisitas harga).
- **Pola waktu:** musiman bulanan, akhir pekan lebih ramai, lonjakan tanggal kembar, kenaikan *Pendidikan* saat tahun ajaran baru (Jul-Agu), *Fashion* dan *Makanan & Minuman* menjelang Lebaran (pertengahan Mar-pertengahan Apr 2024), serta *Elektronik* pada Nov-Des.
- Kombinasi **Produk-Tanggal unik**, dengan *seed* tetap sehingga hasil dapat direproduksi.

````python
def buat_data_penjualan(n=1000, seed=241051019):
    """Dataset operasional penjualan sintetis.
    1 baris = penjualan 1 produk pada 1 tanggal (kombinasi Produk-Tanggal unik).
    Kolom: Produk, Kategori, Harga (Rp), Jumlah Terjual, Tanggal.
    """
    rng = np.random.default_rng(seed)

    # (produk, harga dasar Rp, rata-rata unit terjual per hari)
    katalog = {
        "Elektronik": [("Laptop", 8_500_000, 14), ("Smartphone", 3_200_000, 26),
                       ("Smartwatch", 1_250_000, 30), ("Headphone Bluetooth", 450_000, 60),
                       ("Power Bank", 250_000, 85), ("Mouse Wireless", 150_000, 90)],
        "Pendidikan": [("Buku Pelajaran", 75_000, 120), ("Novel Populer", 95_000, 80),
                       ("Kalkulator Ilmiah", 180_000, 50), ("Tas Ransel Sekolah", 250_000, 45),
                       ("Set Alat Tulis", 55_000, 140)],
        "Fashion": [("Kemeja Pria", 175_000, 55), ("Dress Wanita", 280_000, 40),
                    ("Sepatu Sneakers", 650_000, 28), ("Jaket Hoodie", 320_000, 38),
                    ("Kaos Polos", 85_000, 110)],
        "Makanan & Minuman": [("Kopi Arabika 1 kg", 185_000, 65), ("Teh Celup (Box)", 28_000, 170),
                              ("Mie Instan (Dus)", 115_000, 90), ("Madu Murni", 125_000, 55),
                              ("Keripik Singkong", 20_000, 200)],
        "Rumah Tangga": [("Rice Cooker", 450_000, 32), ("Blender", 380_000, 30),
                         ("Setrika", 250_000, 36), ("Set Panci", 520_000, 22),
                         ("Lampu LED (4 pcs)", 90_000, 95)],
        "Olahraga": [("Matras Yoga", 180_000, 42), ("Dumbbell 10 kg", 350_000, 26),
                     ("Sepatu Lari", 850_000, 20), ("Raket Badminton", 420_000, 24),
                     ("Botol Minum", 95_000, 85)],
        "Kesehatan & Kecantikan": [("Sunscreen", 85_000, 100), ("Serum Wajah", 160_000, 60),
                                   ("Vitamin C (Botol)", 110_000, 90), ("Masker Medis (Box)", 35_000, 150),
                                   ("Parfum", 320_000, 34)],
    }
    produk = pd.DataFrame(
        [(p, k, h, q) for k, items in katalog.items() for p, h, q in items],
        columns=["Produk", "Kategori", "harga_dasar", "qty_dasar"])
    produk["bobot"] = produk["qty_dasar"] ** 0.5   # barang murah lebih sering muncul

    # Pola waktu: musiman bulanan, akhir pekan, dan tanggal kembar (harbolnas)
    tanggal = pd.date_range("2024-01-01", "2024-12-31", freq="D")
    f_bulan = {1: .95, 2: .90, 3: 1.05, 4: 1.25, 5: .95, 6: .95,
               7: 1.10, 8: 1.00, 9: .95, 10: 1.00, 11: 1.20, 12: 1.30}
    kembar = {(3, 3): 1.8, (4, 4): 1.8, (5, 5): 1.8, (6, 6): 2.0, (7, 7): 2.0,
              (8, 8): 2.0, (9, 9): 2.5, (10, 10): 2.5, (11, 11): 4.0, (12, 12): 4.5}

    def bobot_tanggal(t):
        w = f_bulan[t.month]
        w *= 1.25 if t.dayofweek >= 5 else (1.10 if t.dayofweek == 4 else 1.0)
        return w * kembar.get((t.month, t.day), 1.0)

    bt = np.array([bobot_tanggal(t) for t in tanggal])
    bt = bt / bt.sum()

    def boost_kategori(kat, t):
        """Lonjakan permintaan khusus kategori (sekolah, Lebaran, tahun baru, akhir tahun)."""
        if kat == "Pendidikan":
            return 1.8 if t.month in (7, 8) else (1.3 if t.month == 1 else 1.0)
        if kat in ("Fashion", "Makanan & Minuman"):
            return 1.7 if pd.Timestamp("2024-03-15") <= t <= pd.Timestamp("2024-04-14") else 1.0
        if kat == "Olahraga":
            return 1.3 if t.month == 1 else 1.0
        if kat == "Elektronik":
            return 1.4 if t.month in (11, 12) else 1.0
        return 1.0

    baris, terpakai = [], set()
    while len(baris) < n:
        t = tanggal[rng.choice(len(tanggal), p=bt)]
        boost = np.array([boost_kategori(k, t) for k in produk["Kategori"]])
        p = produk["bobot"].to_numpy() * boost
        i = rng.choice(len(produk), p=p / p.sum())
        if (i, t) in terpakai:
            continue
        terpakai.add((i, t))

        dasar = produk.at[i, "harga_dasar"]
        hari_kembar = (t.month, t.day) in kembar
        # Harga: fluktuasi pasar + promo (lebih sering pada tanggal kembar)
        harga = dasar * rng.lognormal(0, 0.06)
        if rng.random() < (0.7 if hari_kembar else 0.08):
            harga *= rng.uniform(0.80, 0.92)
        langkah = 500 if harga < 100_000 else (1_000 if harga < 1_000_000 else 10_000)
        harga = int(round(harga / langkah) * langkah)

        # Jumlah terjual (bilangan bulat): Poisson dengan elastisitas harga dan musiman
        lam = produk.at[i, "qty_dasar"] * boost[i] * f_bulan[t.month] * (dasar / harga) ** 1.3
        lam *= 1.6 if hari_kembar else 1.0
        lam *= 1.10 if t.dayofweek >= 5 else 1.0
        lam *= rng.lognormal(0, 0.25)
        qty = max(1, int(rng.poisson(lam)))
        baris.append((produk.at[i, "Produk"], produk.at[i, "Kategori"], harga, qty, t))

    df = pd.DataFrame(baris, columns=["Produk", "Kategori", "Harga (Rp)", "Jumlah Terjual", "Tanggal"])
    return df.sort_values(["Tanggal", "Produk"]).reset_index(drop=True)


df = buat_data_penjualan(n=1000, seed=241051019)
df.to_csv("data_penjualan_sintetis.csv", index=False)

contoh = df.head(8).copy()
contoh["Tanggal"] = contoh["Tanggal"].dt.strftime("%d/%m/%y")      # format seperti gambar tugas
contoh[HARGA] = contoh[HARGA].map(lambda v: f"{v:,.0f}".replace(",", "."))
print(f"Dataset dibuat: {df.shape[0]:,} baris x {df.shape[1]} kolom".replace(",", ".") + "  ->  data_penjualan_sintetis.csv")
display(contoh)
````

## 4. Validasi Struktur & Kualitas Data

````python
cek = {
    "Tidak ada nilai kosong": df.isna().sum().sum() == 0,
    "Tidak ada baris duplikat (Produk-Tanggal)": not df.duplicated(["Produk", "Tanggal"]).any(),
    "Harga > 0": (df[HARGA] > 0).all(),
    "Jumlah Terjual bilangan bulat >= 1": (df[QTY] >= 1).all(),
    "Tanggal berada pada tahun 2024": df["Tanggal"].dt.year.eq(2024).all(),
    "Tiap produk hanya masuk 1 kategori": df.groupby("Produk", observed=True)["Kategori"].nunique().max() == 1,
    "Jumlah baris = 1.000": len(df) == 1000,
}
for k, v in cek.items():
    print(("[OK]   " if v else "[GAGAL] ") + k)
assert all(cek.values())
````

## 5. Variabel Kategorik (Nominal): Produk & Kategori

Untuk data nominal, ukuran pemusatan yang bermakna hanya **modus** dan **frekuensi**. Rata-rata dan median tidak bermakna. Visualisasi yang tepat: **bar chart** untuk membandingkan jumlah, dan **pie/donut chart** untuk melihat komposisi.

````python
tabel_kat = (df.groupby("Kategori", observed=True)
               .agg(Jumlah_Record=("Produk", "size"), Jumlah_Produk=("Produk", "nunique"),
                    Total_Unit=(QTY, "sum"), Total_Pendapatan=("Pendapatan", "sum"),
                    Median_Harga=(HARGA, "median"))
               .sort_values("Total_Pendapatan", ascending=False))
tabel_kat["% Record"] = tabel_kat["Jumlah_Record"] / tabel_kat["Jumlah_Record"].sum() * 100
tabel_kat["% Pendapatan"] = tabel_kat["Total_Pendapatan"] / tabel_kat["Total_Pendapatan"].sum() * 100

print("Modus Kategori :", df["Kategori"].mode()[0])
print("Modus Produk   :", df["Produk"].mode()[0])
display(tabel_kat.style.format({"Total_Unit": "{:,.0f}", "Total_Pendapatan": lambda v: rp(v),
                                "Median_Harga": lambda v: rp(v), "% Record": "{:.1f}%", "% Pendapatan": "{:.1f}%"}))
````

````python
urut = tabel_kat.sort_values("Jumlah_Record").index
fig, ax = plt.subplots(1, 2, figsize=(14, 5))

# (a) Bar chart: frekuensi record per kategori
nilai = tabel_kat.loc[urut, "Jumlah_Record"]
bars = ax[0].barh(urut.astype(str), nilai, color=BIRU, edgecolor="white")
ax[0].bar_label(bars, labels=[f"{v}  ({v/len(df):.1%})" for v in nilai], padding=4, fontsize=9)
ax[0].set_xlim(0, nilai.max() * 1.22)
ax[0].set_title("Frekuensi record per Kategori (Bar Chart)")
ax[0].set_xlabel("Jumlah record")

# (b) Donut chart: komposisi pendapatan
tp = tabel_kat["Total_Pendapatan"]
w, t, a = ax[1].pie(tp, labels=tp.index.astype(str), autopct=lambda p: f"{p:.1f}%" if p >= 3 else "",
                    colors=PALET_KAT, startangle=90, pctdistance=0.8,
                    wedgeprops=dict(width=0.45, edgecolor="white"), textprops={"fontsize": 9})
for x in a:
    x.set_color("white" if float(x.get_text().strip("%") or 0) > 15 else "#222")
ax[1].set_title("Komposisi pendapatan per Kategori (Donut/Pie Chart)")
ax[1].text(0, 0, f"Total\nRp {tp.sum()/1e9:.1f} M".replace(".", ","), ha="center", va="center", fontsize=10, fontweight="bold")
selesai(fig, "Variabel Nominal: Kategori")
````

````python
prod = (df.groupby(["Produk", "Kategori"], observed=True)
          .agg(Total_Unit=(QTY, "sum"), Total_Pendapatan=("Pendapatan", "sum"), Jumlah_Record=("Produk", "size"))
          .reset_index())
prod = prod[prod["Jumlah_Record"] > 0]

fig, ax = plt.subplots(1, 2, figsize=(14, 5.5))
top_rev = prod.nlargest(10, "Total_Pendapatan").sort_values("Total_Pendapatan")
b = ax[0].barh(top_rev["Produk"].astype(str), top_rev["Total_Pendapatan"] / 1e6, color=BIRU, edgecolor="white")
ax[0].bar_label(b, labels=[f"{v/1e6:,.0f} jt" for v in top_rev["Total_Pendapatan"]], padding=3, fontsize=8.5)
ax[0].set_title("Top 10 Produk berdasarkan pendapatan")
ax[0].set_xlabel("Pendapatan (juta Rp)")
ax[0].set_xlim(0, top_rev["Total_Pendapatan"].max() / 1e6 * 1.18)

top_qty = prod.nlargest(10, "Total_Unit").sort_values("Total_Unit")
b = ax[1].barh(top_qty["Produk"].astype(str), top_qty["Total_Unit"], color="#7CC4E8", edgecolor="white")
ax[1].bar_label(b, padding=3, fontsize=8.5, fmt="{:,.0f}")
ax[1].set_title("Top 10 Produk berdasarkan unit terjual")
ax[1].set_xlabel("Total unit")
ax[1].set_xlim(0, top_qty["Total_Unit"].max() * 1.15)
selesai(fig, "Variabel Nominal: Produk")

print("Top 3 pendapatan :", ", ".join(top_rev["Produk"].astype(str).iloc[::-1].head(3)))
print("Top 3 unit       :", ", ".join(top_qty["Produk"].astype(str).iloc[::-1].head(3)))
print("Catatan: urutan berdasarkan pendapatan (harga x unit) berbeda dari urutan berdasarkan unit, karena harga antar produk sangat timpang.")
````

## 6. Variabel Numerik Kontinu (Rasio): Harga (Rp)

Karena skalanya **rasio**, seluruh statistik deskriptif bermakna (mean, median, simpangan baku, rasio antar nilai). Visualisasi utama: **histogram**. Perhatikan juga **skewness**: harga cenderung miring ke kanan karena sedikit produk sangat mahal.

````python
h = df[HARGA]
desk_h = h.describe(percentiles=[.05, .25, .5, .75, .95]).to_frame("Harga (Rp)")
desk_h.loc["skewness"] = stats.skew(h)
desk_h.loc["kurtosis (excess)"] = stats.kurtosis(h)
desk_h.loc["koef. variasi (%)"] = h.std() / h.mean() * 100
desk_h.loc["modus (pembulatan 10rb)"] = (h // 10_000 * 10_000).mode()[0]

q1, q3 = h.quantile([.25, .75]); iqr = q3 - q1
batas_atas = q3 + 1.5 * iqr
n_out = (h > batas_atas).sum()
display(desk_h)
print(f"IQR global = {rp(iqr)} | batas atas pencilan = {rp(batas_atas)} | jumlah 'pencilan' = {n_out} ({n_out/len(h):.1%})")
print("Catatan: pencilan global berasal dari produk berharga tinggi yang valid (bukan kesalahan data); lihat tabel per kategori di bawah.")

# Pencilan dihitung per kategori (lebih adil)
def n_outlier(s):
    a, b = s.quantile([.25, .75]); k = b - a
    return ((s < a - 1.5 * k) | (s > b + 1.5 * k)).sum()
display(df.groupby("Kategori", observed=True)[HARGA].agg(["count", "min", "median", "mean", "max", n_outlier])
          .rename(columns={"n_outlier": "pencilan_dlm_kategori"}).style.format(
              {"min": rp, "median": rp, "mean": rp, "max": rp}))
````

````python
fig, ax = plt.subplots(2, 2, figsize=(14, 9))

# (a) Histogram + KDE skala asli
sns.histplot(h, bins=40, kde=True, kde_kws={"cut": 0}, color=ORANYE, edgecolor="white", ax=ax[0, 0])
ax[0, 0].axvline(h.mean(), color="#C0392B", ls="--", lw=1.4, label=f"Mean {rp(h.mean())}")
ax[0, 0].axvline(h.median(), color="#27AE60", ls="--", lw=1.4, label=f"Median {rp(h.median())}")
ax[0, 0].xaxis.set_major_formatter(FuncFormatter(fmt_juta))
ax[0, 0].set(title="Histogram Harga (skala asli) - miring ke kanan", xlabel="Harga (juta Rp)", ylabel="Frekuensi")
ax[0, 0].legend(fontsize=8.5)

# (b) Histogram skala log10
sns.histplot(np.log10(h), bins=40, kde=True, color=ORANYE, edgecolor="white", ax=ax[0, 1])
ax[0, 1].set(title="Histogram log10(Harga) - sebaran lebih mudah dibaca", xlabel="log10(Harga Rp)", ylabel="Frekuensi")

# (c) Boxplot per kategori (log)
urut_h = df.groupby("Kategori", observed=True)[HARGA].median().sort_values().index.astype(str)
sns.boxplot(data=df.assign(Kategori=df["Kategori"].astype(str)), y="Kategori", x=HARGA, order=urut_h,
            color=ORANYE, fliersize=2.5, linewidth=1, ax=ax[1, 0])
ax[1, 0].set_xscale("log")
ax[1, 0].set(title="Sebaran Harga per Kategori (boxplot, skala log)", xlabel="Harga (Rp, skala log)", ylabel="")

# (d) ECDF
sns.ecdfplot(h, color=ORANYE, lw=2, ax=ax[1, 1])
for p in (.5, .9):
    v = h.quantile(p)
    ax[1, 1].hlines(p, h.min(), v, color="#999", ls=":"); ax[1, 1].vlines(v, 0, p, color="#999", ls=":")
    ax[1, 1].annotate(f"P{int(p*100)} = {rp(v)}", (v, p), xytext=(8, -14), textcoords="offset points", fontsize=8.5)
ax[1, 1].set_xscale("log")
ax[1, 1].xaxis.set_major_formatter(FuncFormatter(fmt_rupiah_singkat))
ax[1, 1].set(title="ECDF Harga (proporsi kumulatif, skala log)", xlabel="Harga (Rp, skala log)", ylabel="Proporsi")
selesai(fig, "Variabel Numerik Kontinu (Rasio): Harga (Rp)")
````

````python
# Variasi harga di dalam produk yang sama (efek fluktuasi pasar dan promo)
var_prod = (df.groupby("Produk", observed=True)[HARGA].agg(["mean", "std", "min", "max"])
              .assign(CV_persen=lambda d: d["std"] / d["mean"] * 100).sort_values("CV_persen", ascending=False))
display(var_prod.head(8).style.format({"mean": rp, "std": rp, "min": rp, "max": rp, "CV_persen": "{:.1f}%"}))
print(f"Rata-rata koefisien variasi harga dalam satu produk = {var_prod['CV_persen'].mean():.1f}%  (jauh lebih kecil daripada CV global {h.std()/h.mean()*100:.0f}%)")
````

## 7. Variabel Numerik Diskrit (Rasio): Jumlah Terjual

Nilainya **bilangan bulat hasil penghitungan** dengan nol absolut, sehingga statistik rasio berlaku. Visualisasi yang cocok: histogram/bar dengan lebar kelas bulat. Perlu dicek **overdispersion** (varians jauh lebih besar dari rata-rata), yang wajar karena produk yang berbeda memiliki laju penjualan berbeda.

````python
q = df[QTY]
desk_q = q.describe(percentiles=[.05, .25, .5, .75, .95]).to_frame("Jumlah Terjual")
desk_q.loc["skewness"] = stats.skew(q)
desk_q.loc["kurtosis (excess)"] = stats.kurtosis(q)
desk_q.loc["varians"] = q.var()
desk_q.loc["varians / mean"] = q.var() / q.mean()
desk_q.loc["modus"] = q.mode()[0]
display(desk_q)

print(f"Rasio varians/mean = {q.var()/q.mean():.1f}  -> overdispersi (>1): penjualan antar produk sangat heterogen.")
````

````python
fig, ax = plt.subplots(2, 2, figsize=(14, 9))

# (a) Histogram nilai bulat
lebar = 10
sns.histplot(q, binwidth=lebar, color=ORANYE, edgecolor="white", ax=ax[0, 0])
ax[0, 0].axvline(q.mean(), color="#C0392B", ls="--", label=f"Mean {q.mean():.1f}")
ax[0, 0].axvline(q.median(), color="#27AE60", ls="--", label=f"Median {q.median():.0f}")
ax[0, 0].set(title=f"Distribusi Jumlah Terjual (lebar kelas = {lebar} unit)", xlabel="Unit terjual", ylabel="Frekuensi")
ax[0, 0].legend(fontsize=8.5)

# (b) Boxplot per kategori
urut_q = df.groupby("Kategori", observed=True)[QTY].median().sort_values().index.astype(str)
sns.boxplot(data=df.assign(Kategori=df["Kategori"].astype(str)), y="Kategori", x=QTY, order=urut_q,
            color=ORANYE, fliersize=2.5, linewidth=1, ax=ax[0, 1])
ax[0, 1].set(title="Sebaran Jumlah Terjual per Kategori", xlabel="Unit terjual", ylabel="")

# (c) Rata-rata unit per kategori + 95% CI
sns.barplot(data=df.assign(Kategori=df["Kategori"].astype(str)), y="Kategori", x=QTY, order=urut_q[::-1],
            color="#F8C382", errorbar=("ci", 95), capsize=.25, ax=ax[1, 0])
ax[1, 0].set(title="Rata-rata unit per record (95% CI)", xlabel="Rata-rata unit", ylabel="")

# (d) Total unit per bulan (bar - data bulat)
per_bulan = df.groupby("Bulan_Nama", observed=True)[QTY].sum()
bb = ax[1, 1].bar(per_bulan.index.astype(str), per_bulan.values, color=ORANYE, edgecolor="white")
ax[1, 1].bar_label(bb, fmt="{:,.0f}", fontsize=7.5, padding=2, rotation=90)
ax[1, 1].set_ylim(0, per_bulan.max() * 1.18)
ax[1, 1].set(title="Total unit terjual per bulan", xlabel="Bulan", ylabel="Unit")
selesai(fig, "Variabel Numerik Diskrit (Rasio): Jumlah Terjual")
````

## 8. Variabel Waktu (Interval): Tanggal

Pada skala **interval**, selisih antar tanggal bermakna (misalnya 30 hari), tetapi tanggal tidak punya nol absolut, sehingga rasio antar tanggal tidak bermakna. Visualisasi yang tepat adalah **line chart** untuk melihat **tren, musiman, dan lonjakan**.

````python
harian = (df.groupby("Tanggal")[["Pendapatan", QTY]].sum()
            .reindex(pd.date_range("2024-01-01", "2024-12-31"), fill_value=0))
harian["MA7"] = harian["Pendapatan"].rolling(7, center=True, min_periods=1).mean()

fig, ax = plt.subplots(figsize=(14, 5))
ax.plot(harian.index, harian["Pendapatan"] / 1e6, color="#B9BCC2", lw=.9, label="Pendapatan harian")
ax.plot(harian.index, harian["MA7"] / 1e6, color=ABU, lw=2.4, label="Rata-rata bergerak 7 hari")
for tgl, brs in harian["Pendapatan"].nlargest(3).items():
    ax.annotate(f"{tgl:%d %b}\n{brs/1e6:,.0f} jt", (tgl, brs / 1e6), xytext=(0, 12), textcoords="offset points",
                ha="center", fontsize=8.5, arrowprops=dict(arrowstyle="-", color="#888"))
ax.set(title="Tren pendapatan harian 2024 (Line Chart)", xlabel="Tanggal", ylabel="Pendapatan (juta Rp)")
ax.set_ylim(0, harian["Pendapatan"].max() / 1e6 * 1.2)
ax.legend(loc="upper left")
selesai(fig, "Variabel Waktu (Interval): Tren Harian")
````

````python
bulanan = (df.groupby("Bulan_Nama", observed=True)
             .agg(Pendapatan=("Pendapatan", "sum"), Unit=(QTY, "sum"), Record=("Produk", "size")))
bulanan["Pertumbuhan_MoM_%"] = bulanan["Pendapatan"].pct_change() * 100

fig, ax = plt.subplots(1, 2, figsize=(14, 4.8))
ax[0].plot(bulanan.index.astype(str), bulanan["Pendapatan"] / 1e6, marker="o", color=ABU, lw=2.4)
pk = bulanan["Pendapatan"].idxmax()
ax[0].annotate(f"Puncak: {pk}", (list(bulanan.index).index(pk), bulanan.loc[pk, "Pendapatan"] / 1e6),
               xytext=(0, 12), textcoords="offset points", ha="center", fontsize=9)
ax[0].set(title="Pendapatan bulanan", xlabel="Bulan", ylabel="Pendapatan (juta Rp)")
ax[0].set_ylim(0, bulanan["Pendapatan"].max() / 1e6 * 1.2)

ax[1].plot(bulanan.index.astype(str), bulanan["Unit"], marker="s", color="#4F5560", lw=2.4)
ax[1].set(title="Unit terjual bulanan", xlabel="Bulan", ylabel="Unit")
ax[1].set_ylim(0, bulanan["Unit"].max() * 1.2)
selesai(fig, "Tren Bulanan")
display(bulanan.style.format({"Pendapatan": rp, "Unit": "{:,.0f}", "Pertumbuhan_MoM_%": "{:+.1f}%"}, na_rep="-"))
````

````python
kat_bulan = df.pivot_table(index="Bulan_Nama", columns="Kategori", values="Pendapatan", aggfunc="sum", observed=True)

fig, ax = plt.subplots(1, 2, figsize=(15, 5.2), gridspec_kw={"width_ratios": [1.15, 1]})
for i, kat in enumerate(kat_bulan.columns):
    ax[0].plot(kat_bulan.index.astype(str), kat_bulan[kat] / 1e6, marker="o", ms=4, lw=1.8, color=sns.color_palette("tab10")[i], label=str(kat))
ax[0].set_yscale("log")
ax[0].set(title="Tren bulanan per Kategori (skala log)", xlabel="Bulan", ylabel="Pendapatan (juta Rp, log)")
ax[0].legend(fontsize=8, ncol=2)

sns.heatmap(kat_bulan.T / 1e6, annot=True, fmt=".0f", cmap="Greys", cbar_kws={"label": "juta Rp"}, linewidths=.5, ax=ax[1], annot_kws={"fontsize": 7.5})
ax[1].set(title="Heatmap pendapatan Kategori x Bulan (juta Rp)", xlabel="", ylabel="")
selesai(fig, "Musiman per Kategori")
````

````python
hari = df.groupby("Hari_Nama", observed=True).agg(Record=("Produk", "size"), Pendapatan=("Pendapatan", "sum"), Unit=(QTY, "sum"))
hari["Rata2_Unit_per_record"] = hari["Unit"] / hari["Record"]

fig, ax = plt.subplots(1, 2, figsize=(14, 4.5))
warna = [ABU] * 5 + ["#4F5560"] * 2
b = ax[0].bar(hari.index.astype(str), hari["Record"], color=warna, edgecolor="white")
ax[0].bar_label(b, padding=2, fontsize=9)
ax[0].set(title="Jumlah record per hari dalam seminggu", ylabel="Jumlah record")
ax[0].set_ylim(0, hari["Record"].max() * 1.15)
b = ax[1].bar(hari.index.astype(str), hari["Rata2_Unit_per_record"], color=warna, edgecolor="white")
ax[1].bar_label(b, fmt="%.0f", padding=2, fontsize=9)
ax[1].set(title="Rata-rata unit per record per hari", ylabel="Unit")
ax[1].set_ylim(0, hari["Rata2_Unit_per_record"].max() * 1.15)
selesai(fig, "Pola Mingguan (akhir pekan = abu tua)")

kembar = df[df["Tanggal"].dt.strftime("%m-%d").isin(["11-11", "12-12", "10-10", "09-09"])]
print(f"Record pada tanggal kembar 9.9 / 10.10 / 11.11 / 12.12 = {len(kembar)}, "
      f"rata-rata unit/record = {kembar[QTY].mean():.0f} vs hari biasa = {df.drop(kembar.index)[QTY].mean():.0f}")
````

## 9. Hubungan Antar Variabel

Karena distribusi harga sangat miring, korelasi **Spearman** (berbasis peringkat) lebih tepat daripada Pearson. Analisis dilakukan pada dua tingkat: antar semua produk, dan di dalam produk yang sama (untuk melihat pengaruh promo/diskon).

````python
rho_all, p_all = stats.spearmanr(df[HARGA], df[QTY])

# Harga & unit relatif terhadap median produk masing-masing (efek promo di dalam produk)
med = df.groupby("Produk", observed=True)[[HARGA, QTY]].transform("median")
df["Harga_rel"] = df[HARGA] / med[HARGA]
df["Qty_rel"] = df[QTY] / med[QTY]
rho_rel, p_rel = stats.spearmanr(df["Harga_rel"], df["Qty_rel"])

fig, ax = plt.subplots(1, 2, figsize=(14, 5.3))
sns.scatterplot(data=df.assign(Kategori=df["Kategori"].astype(str)), x=HARGA, y=QTY, hue="Kategori",
                palette="tab10", s=22, alpha=.7, ax=ax[0])
ax[0].set_xscale("log"); ax[0].set_yscale("log")
ax[0].set(title=f"Harga vs Jumlah Terjual (log-log)\nSpearman rho = {rho_all:.2f}", xlabel="Harga (Rp, log)", ylabel="Unit (log)")
ax[0].legend(fontsize=7.5, loc="upper right")

ax[1].scatter(df["Harga_rel"], df["Qty_rel"], s=18, alpha=.55, color=ORANYE, edgecolor="none")
tren = df.groupby(pd.qcut(df["Harga_rel"], 10), observed=True)[["Harga_rel", "Qty_rel"]].median()
ax[1].plot(tren["Harga_rel"], tren["Qty_rel"], color="#C0392B", lw=2.4, marker="o", label="Median per desil harga")
ax[1].legend(fontsize=8.5)
ax[1].axvline(1, color="#999", ls=":"); ax[1].axhline(1, color="#999", ls=":")
ax[1].set(title=f"Di dalam produk: harga relatif vs unit relatif\nSpearman rho = {rho_rel:.2f}",
          xlabel="Harga / median harga produk", ylabel="Unit / median unit produk")
selesai(fig, "Hubungan Harga dan Jumlah Terjual")

print(f"Antar produk  : rho = {rho_all:+.3f} (p = {p_all:.1e})  -> barang mahal cenderung terjual lebih sedikit")
print(f"Dalam produk  : rho = {rho_rel:+.3f} (p = {p_rel:.1e})  -> saat harga turun (promo), unit terjual naik")
````

````python
num = df[[HARGA, QTY, "Pendapatan"]]
fig, ax = plt.subplots(1, 2, figsize=(13, 4.8), gridspec_kw={"width_ratios": [1, 1.1]})
sns.heatmap(num.corr(method="spearman"), annot=True, fmt=".2f", cmap="RdBu_r", vmin=-1, vmax=1, ax=ax[0], square=True)
ax[0].set_title("Matriks korelasi Spearman")

order = df.groupby("Kategori", observed=True)["Pendapatan"].median().sort_values().index.astype(str)
sns.boxplot(data=df.assign(Kategori=df["Kategori"].astype(str)), y="Kategori", x="Pendapatan", order=order, color=ORANYE, fliersize=2, ax=ax[1])
ax[1].set_xscale("log")
ax[1].set(title="Pendapatan per record per Kategori (log)", xlabel="Pendapatan (Rp, log)", ylabel="")
selesai(fig, "Korelasi & Pendapatan")

# Uji perbedaan antar kategori (non-parametrik)
for var in (HARGA, QTY, "Pendapatan"):
    grup = [g[var].values for _, g in df.groupby("Kategori", observed=True)]
    H, p = stats.kruskal(*grup)
    print(f"Kruskal-Wallis {var:15s}: H = {H:8.1f}, p = {p:.1e}  -> {'ada' if p < .05 else 'tidak ada'} perbedaan antar kategori")
````

````python
kat_top = tabel_kat.index[0]
prod_top_rev = prod.nlargest(1, "Total_Pendapatan").iloc[0]
prod_top_qty = prod.nlargest(1, "Total_Unit").iloc[0]
bulan_top = bulanan["Pendapatan"].idxmax(); bulan_low = bulanan["Pendapatan"].idxmin()
hari_top = harian["Pendapatan"].idxmax()
share_wkend = df.loc[df["Akhir_Pekan"], "Pendapatan"].sum() / df["Pendapatan"].sum()
rec_wkend = df["Akhir_Pekan"].mean()

temuan = f"""
**Nominal (Produk & Kategori)**
- Kategori dengan pendapatan terbesar: **{kat_top}** ({tabel_kat.loc[kat_top, '% Pendapatan']:.1f}% dari total), walau hanya {tabel_kat.loc[kat_top, '% Record']:.1f}% dari jumlah record.
- Produk dengan pendapatan tertinggi: **{prod_top_rev['Produk']}** ({rp(prod_top_rev['Total_Pendapatan'])}); produk dengan unit terbanyak: **{prod_top_qty['Produk']}** ({prod_top_qty['Total_Unit']:,} unit).

**Numerik kontinu (Harga)**
- Harga sangat miring ke kanan (skewness = {stats.skew(df[HARGA]):.1f}); median {rp(df[HARGA].median())} jauh di bawah mean {rp(df[HARGA].mean())}. Skala log membuat distribusinya lebih simetris.

**Numerik diskrit (Jumlah Terjual)**
- Rata-rata {df[QTY].mean():.1f} unit per record (median {df[QTY].median():.0f}), dengan overdispersi (varians/mean = {df[QTY].var()/df[QTY].mean():.0f}) karena laju penjualan antar produk berbeda jauh.

**Waktu (Tanggal)**
- Bulan tertinggi: **{bulan_top}**, terendah: **{bulan_low}**. Hari dengan pendapatan tertinggi: **{hari_top:%d/%m/%Y}** (kemungkinan efek tanggal kembar/harbolnas).
- Akhir pekan menyumbang {share_wkend:.1%} pendapatan dari {rec_wkend:.1%} record.

**Hubungan**
- Antar produk, harga dan unit berkorelasi negatif (rho = {rho_all:.2f}); di dalam produk yang sama, diskon meningkatkan unit terjual (rho = {rho_rel:.2f}).
"""
display(Markdown(temuan))
````
